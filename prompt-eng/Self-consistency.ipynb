```python
from _pipeline import create_payload, model_req

# (1) Define the simulated inbound message
# Note: This message represents an inquiry about the bot’s requirements.
MESSAGE = "Startup Mentor Bot Requirements"

# (2) Adjust the Prompt Engineering Technique to be applied
SELF_CONSISTENCY_PROMPT = \
f"""
To ensure robust and consistent requirement analysis, let’s derive the specifications multiple times and synthesize the most repeated insights:

Attempt 1:
A startup mentor bot should provide:
- Feedback on startup pitches and ideas.
- Access to startup checklists and guides.
- Integration with scheduling tools like Calendly.
- Lightweight storage for user preferences.

Attempt 2:
The bot must:
- Offer tailored startup advice via conversational flow.
- Connect to APIs like Crunchbase for market research.
- Support document uploads (e.g., pitch decks, business plans).
- Prioritize secure access with mentor/mentee roles.

Attempt 3:
Core features should include:
- Step-by-step startup idea validation.
- Role-based guidance (early-stage, fundraising, scaling).
- Integration with Notion/Google Docs for live feedback.
- Efficient caching and session state maintenance.

From these attempts, the key consistent requirements are:

1. **Core Functionalities**
   - Startup idea evaluation and structured guidance.
   - Personalized mentorship based on founder stage.
   - Resource checklists and growth tools.

2. **User Interaction**
   - Conversational advice delivery via commands.
   - Upload and feedback interface for pitch decks.
   - Role-based mentoring (e.g., beginner, investor-ready).

3. **Integration & Extensibility**
   - APIs (e.g., Crunchbase, Notion, Google Slides).
   - Calendly or similar scheduling tool integration.
   - Modular plugin support for advanced workflows.

4. **Performance & Optimization**
   - Session management for ongoing mentorship threads.
   - Responsive, low-latency message handling.
   - Efficient use of AI context across interactions.

5. **Security & Permissions**
   - Mentor-mentee role separation.
   - Logging of sessions with anonymized data.
   - Access restrictions for confidential materials.

Extract the consistent requirements from the given topic: {MESSAGE}. Provide a structured response.
A:
"""

PROMPT = SELF_CONSISTENCY_PROMPT

# (3) Configure the payload (optional)
payload = create_payload(
    target="ollama",
    model="llama3.2",
    prompt=PROMPT,
    temperature=0.7,
    num_ctx=1024,
    num_predict=512
)

# Send to the model
time, response = model_req(payload=payload)
print(response)
if time:
    print(f"Time taken: {time}s")
```
