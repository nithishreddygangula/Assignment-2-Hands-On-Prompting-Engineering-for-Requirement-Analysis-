```python
from _pipeline import create_payload, model_req

# (1) Define the simulated inbound message
# Note: This message represents an inquiry about the bot’s requirements.
MESSAGE = "Startup Mentor Bot"

# (2) Implement Few-Shot Prompting Technique
FEWSHOT_PROMPT = \
f"""
Example 1:
Q: What are the core requirements for a productivity assistant bot?
A:
- Create and manage task lists with due dates.
- Send reminders and productivity tips.
- Integrate with calendar tools.
- Support daily/weekly planning.

Example 2:
Q: What are the core requirements for a nutrition tracking bot?
A:
- Log daily meals and calories.
- Offer dietary recommendations based on user goals.
- Provide nutritional info from a food database.
- Visualize weekly nutrition trends.

Now, based on the above examples, list only the essential and core requirements for a {MESSAGE} in a structured format.
"""

# (3) Configure the payload for generating requirements
payload = create_payload(
    target="ollama",
    model="llama3.2",
    prompt=FEWSHOT_PROMPT,
    temperature=0.4,     # Ensuring focused and structured response
    num_ctx=512,         # Allowing adequate context
    num_predict=256      # Keeping response concise
)

# Send out to the model
time, response = model_req(payload=payload)
print("Generated Requirements:")
print(response)
if time:
    print(f'Time taken: {time}s')
```
