```python
from _pipeline import create_payload, model_req

# (1) Define the simulated inbound message
# Note: This message represents an inquiry about the bot’s requirements.
MESSAGE = "Startup Mentor Bot Requirements"

# (2) Refined Prompt for High-Quality Structured Output
REFINED_PROMPT = \
f"""
You are an AI designed for product specification and requirement analysis. Generate a well-structured breakdown of functional and non-functional requirements for a Discord bot aimed at helping startup founders and entrepreneurs.

### **Requirements:**
- **Core Features:** Idea validation guidance, pitch deck feedback, startup checklists, and mentorship prompts.
- **User Interaction:** Command-based questions, feedback summaries, and intuitive guidance workflows.
- **Integrations:** APIs such as Crunchbase, Notion, Google Slides, or Calendly for startup management.
- **Performance:** Fast response times, scalable user support, and smooth interaction for long sessions.
- **Security:** User-specific recommendation logs, minimal data retention, and role-restricted expert access.

Generate a structured list of requirements for: **{MESSAGE}**
A:
"""

PROMPT = REFINED_PROMPT

# (3) Optimized Payload Configuration
payload = create_payload(
    target="ollama",
    model="llama3.2",
    prompt=PROMPT,
    temperature=0.3,         # Ensuring precise and concise answers
    num_ctx=1024,            # Maximizing context window
    num_predict=500,         # Allowing sufficient output
    top_p=0.8,               # Keeping responses focused
    frequency_penalty=0.3,   # Reducing redundancy
    presence_penalty=0.4     # Encouraging diverse response content
)

# Send out to the model
time, response = model_req(payload=payload)
print(response)
if time:
    print(f"Time taken: {time}s")
```
