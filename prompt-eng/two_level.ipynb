```python
from _pipeline import create_payload, model_req

# (1) Define the simulated inbound message
# This message reflects a request for startup mentorship bot requirements.
MESSAGE = "Startup Mentor Bot Requirements"

# (2) Implement Two-Level Prompting Technique
LEVEL_ONE_PROMPT = \
f"""
You are an AI assistant specializing in startup-focused requirement analysis. Generate a compact list of insightful questions to extract critical requirements for a Discord bot designed to mentor startup founders.

Focus on:
- Core features (idea validation, pitch review, market research tips)
- User interaction (conversational advice, checklists, resource links)
- Integrations (Crunchbase, Notion, Google Docs, Calendly)
- Performance (fast feedback, multi-session handling, scalability)
- Security (confidential data handling, access control)

List only 5–7 strategic questions.
A:
"""

# First Level Response Extraction
payload_level_one = create_payload(
    target="ollama",
    model="llama3.2",
    prompt=LEVEL_ONE_PROMPT,
    temperature=0.5,
    num_ctx=512,
    num_predict=256
)

time, questions_response = model_req(payload=payload_level_one)
print("Generated Questions:")
print(questions_response)
if time:
    print(f"Time taken: {time}s")

# (3) Use Second-Level Prompt to Extract Final Requirements
LEVEL_TWO_PROMPT = \
f"""
Using the following strategic questions, generate a clear and structured list of functional and non-functional requirements for a Startup Mentor Discord Bot:

{questions_response}

Limit the output to the most essential requirements.
A:
"""

# (4) Configure the payload for the second level request
payload_level_two = create_payload(
    target="ollama",
    model="llama3.2",
    prompt=LEVEL_TWO_PROMPT,
    temperature=0.5,
    num_ctx=512,
    num_predict=256
)

time, response = model_req(payload=payload_level_two)
print("Final Requirements:")
print(response)
if time:
    print(f"Time taken: {time}s")
```
